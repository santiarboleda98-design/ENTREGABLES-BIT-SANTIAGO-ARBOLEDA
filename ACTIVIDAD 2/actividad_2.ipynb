```python
# Identificar los pesos negativos y convertirlos en NaN
df_clean.loc[
    df_clean["Weight kg"] < 0,
    "Weight kg"
] = np.nan

print("Pesos faltantes:", df_clean["Weight kg"].isnull().sum())
```

```python
# Calcular la mediana del peso por tipo de animal
medianas_por_animal = df_clean.groupby("Animal type")["Weight kg"].transform("median")

# Completar los pesos faltantes con la mediana de su categoría
df_clean["Weight kg"] = df_clean["Weight kg"].fillna(medianas_por_animal)

# Verificar el resultado
print("Pesos faltantes:", df_clean["Weight kg"].isnull().sum())
print("Pesos negativos:", (df_clean["Weight kg"] < 0).sum())
```

Se identificaron tres registros con pesos negativos, todos correspondientes a la categoría red squirrel. Debido a que los pesos negativos no son físicamente válidos, se reemplazaron temporalmente por valores faltantes (NaN). Posteriormente, se imputaron utilizando la mediana del peso de cada tipo de animal. Después de la transformación, no quedaron pesos negativos ni valores faltantes en la columna Weight kg.

```python
# 1. Cantidad de registros por tipo de animal
observaciones_por_animal = df_clean.groupby("Animal type").size()
print("Registros por tipo de animal:")
print(observaciones_por_animal)

# 2. Peso promedio por tipo de animal
peso_promedio_animal = (
    df_clean.groupby("Animal type")["Weight kg"].mean()
)
print("\nPeso promedio por tipo de animal:")
print(peso_promedio_animal)
```

```python
# 3. Peso promedio por país
peso_promedio_pais = (
    df_clean.groupby("Country")["Weight kg"].mean()
)

print("Peso promedio por país:")
print(peso_promedio_pais)

# 4. Cantidad de registros por país
observaciones_por_pais = (
    df_clean.groupby("Country").size()
)

print("\nRegistros por país:")
print(observaciones_por_pais.sort_values(ascending=False))
```

3. Distribución geográfica de las observaciones

Polonia cuenta con 192 registros, la mayor cantidad entre los países identificados. Austria presenta 78, la menor cantidad. Estos resultados describen la distribución de las observaciones disponibles y no permiten determinar por sí solos la abundancia real de animales en cada país.
